```python

# ============================================================
# B.Tech Mathematics Mini Project
# ROBOT ARM POSITION CALCULATOR (F2)
# 2-Link Robot Arm in 2D
# Google Colab - Single Cell Version
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
import pandas as pd

# ============================================================
# 1. USER CONFIGURATION / INPUTS
# ============================================================

# Link lengths
L1 = 5.0
L2 = 3.0

# Desired end-effector position
# Change these values for your own problem
TARGET_X = 6.0
TARGET_Y = 3.0

# Choose inverse-kinematics solution:
# "elbow_up" or "elbow_down"
SOLUTION = "elbow_up"

# Run the benchmark test also
RUN_BENCHMARK = True


# ============================================================
# 2. MATHEMATICAL FUNCTIONS
# ============================================================

def rotation_matrix(theta_deg):
    """
    Create a 2D rotation matrix.

    R(theta) =
        [ cos(theta)  -sin(theta) ]
        [ sin(theta)   cos(theta) ]

    theta is given in degrees.
    """
    theta = np.radians(theta_deg)

    return np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])


def forward_kinematics(L1, L2, theta1_deg, theta2_deg):
    """
    Calculate the position of the end of a 2-link planar robot.

    The second joint angle is relative to the first link.

    x = L1*cos(theta1) + L2*cos(theta1 + theta2)
    y = L1*sin(theta1) + L2*sin(theta1 + theta2)
    """

    theta1 = np.radians(theta1_deg)
    theta2 = np.radians(theta2_deg)

    # Position of joint 1
    x1 = L1 * np.cos(theta1)
    y1 = L1 * np.sin(theta1)

    # Position of end effector
    x2 = x1 + L2 * np.cos(theta1 + theta2)
    y2 = y1 + L2 * np.sin(theta1 + theta2)

    return x1, y1, x2, y2


def inverse_kinematics(L1, L2, x, y, solution="elbow_up"):
    """
    Calculate joint angles for a desired end position.

    Uses the standard 2-link planar inverse-kinematics equations.

    cos(theta2) =
        (x^2 + y^2 - L1^2 - L2^2) / (2*L1*L2)

    theta1 =
        atan2(y,x) - atan2(L2*sin(theta2),
                           L1 + L2*cos(theta2))
    """

    distance_squared = x**2 + y**2

    # Check whether target is reachable
    min_reach = abs(L1 - L2)
    max_reach = L1 + L2
    distance = np.sqrt(distance_squared)

    if distance > max_reach + 1e-10:
        raise ValueError(
            f"Target ({x}, {y}) is outside the robot's maximum reach "
            f"of {max_reach:.2f}."
        )

    if distance < min_reach - 1e-10:
        raise ValueError(
            f"Target ({x}, {y}) is inside the robot's minimum reach "
            f"of {min_reach:.2f}."
        )

    # Inverse-kinematics equation for theta2
    cos_theta2 = (
        distance_squared - L1**2 - L2**2
    ) / (2 * L1 * L2)

    # Numerical safety against tiny floating-point errors
    cos_theta2 = np.clip(cos_theta2, -1.0, 1.0)

    theta2_abs = np.arccos(cos_theta2)

    if solution.lower() == "elbow_down":
        theta2 = -theta2_abs
    else:
        theta2 = theta2_abs

    # Calculate theta1
    theta1 = (
        np.arctan2(y, x)
        - np.arctan2(
            L2 * np.sin(theta2),
            L1 + L2 * np.cos(theta2)
        )
    )

    return np.degrees(theta1), np.degrees(theta2)


# ============================================================
# 3. SOLVE INVERSE KINEMATICS
# ============================================================

theta1, theta2 = inverse_kinematics(
    L1,
    L2,
    TARGET_X,
    TARGET_Y,
    SOLUTION
)

# Forward calculation using the obtained angles
x1, y1, calculated_x, calculated_y = forward_kinematics(
    L1,
    L2,
    theta1,
    theta2
)

error = np.sqrt(
    (TARGET_X - calculated_x)**2 +
    (TARGET_Y - calculated_y)**2
)


# ============================================================
# 4. DISPLAY MATHEMATICAL MODEL
# ============================================================

R1 = rotation_matrix(theta1)
R2 = rotation_matrix(theta2)

# Overall rotation transformation
R_total = R2 @ R1

print("=" * 65)
print("       ROBOT ARM POSITION CALCULATOR (F2)")
print("=" * 65)

print("\nASSUMPTION")
print("-" * 65)
print("A 2-link planar robot arm is used.")
print("Angles are measured counter-clockwise.")
print("theta2 is the relative angle of link 2 with respect to link 1.")

print("\nINPUTS")
print("-" * 65)
print(f"Link 1 length       : {L1:.2f}")
print(f"Link 2 length       : {L2:.2f}")
print(f"Target X            : {TARGET_X:.2f}")
print(f"Target Y            : {TARGET_Y:.2f}")
print(f"IK solution         : {SOLUTION}")

print("\nROTATION MATRIX R1")
print("-" * 65)
print(np.round(R1, 4))

print("\nROTATION MATRIX R2")
print("-" * 65)
print(np.round(R2, 4))

print("\nTOTAL ROTATION MATRIX")
print("-" * 65)
print("R_total = R2 × R1")
print(np.round(R_total, 4))

print("\nINVERSE KINEMATICS RESULT")
print("-" * 65)
print(f"Theta 1 = {theta1:.4f} degrees")
print(f"Theta 2 = {theta2:.4f} degrees")

print("\nFORWARD KINEMATICS VERIFICATION")
print("-" * 65)
print(f"Joint 1 position     : ({x1:.4f}, {y1:.4f})")
print(f"Calculated end point : ({calculated_x:.4f}, {calculated_y:.4f})")
print(f"Desired end point    : ({TARGET_X:.4f}, {TARGET_Y:.4f})")
print(f"Position error       : {error:.10f}")

if error < 1e-6:
    print("\nSTATUS: ✓ Target reached accurately")
else:
    print("\nSTATUS: Target calculation completed")


# ============================================================
# 5. RESULT TABLE
# ============================================================

result_table = pd.DataFrame({
    "Parameter": [
        "Link 1 Length",
        "Link 2 Length",
        "Target X",
        "Target Y",
        "Joint Angle θ1",
        "Joint Angle θ2",
        "Calculated X",
        "Calculated Y",
        "Position Error"
    ],
    "Value": [
        L1,
        L2,
        TARGET_X,
        TARGET_Y,
        theta1,
        theta2,
        calculated_x,
        calculated_y,
        error
    ]
})

print("\nRESULT TABLE")
display(result_table.round(6))


# ============================================================
# 6. STATIC ROBOT ARM VISUALIZATION
# ============================================================

fig, ax = plt.subplots(figsize=(8, 8))

# Robot arm
ax.plot(
    [0, x1, calculated_x],
    [0, y1, calculated_y],
    marker="o",
    linewidth=5,
    markersize=10,
    label="Robot Arm"
)

# Target point
ax.scatter(
    TARGET_X,
    TARGET_Y,
    s=180,
    marker="*",
    label="Desired Target"
)

# Joint labels
ax.text(0, 0, "  Base", fontsize=11)
ax.text(x1, y1, "  Joint 1", fontsize=11)
ax.text(calculated_x, calculated_y, "  End Effector", fontsize=11)

# Draw target line
ax.plot(
    [calculated_x, TARGET_X],
    [calculated_y, TARGET_Y],
    linestyle="--",
    linewidth=2,
    label="Position Error"
)

limit = L1 + L2 + 1

ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)

ax.set_aspect("equal")
ax.grid(True, alpha=0.3)

ax.set_xlabel("X Position")
ax.set_ylabel("Y Position")
ax.set_title("2-Link Robot Arm — Position Calculator")

ax.legend()

plt.show()


# ============================================================
# 7. BENCHMARK TEST DATASET
# ============================================================

if RUN_BENCHMARK:

    print("\n" + "=" * 65)
    print("              BENCHMARK TEST DATASET")
    print("=" * 65)

    benchmark_data = [
        # L1, L2, X, Y
        (5.0, 3.0, 6.0, 3.0),
        (4.0, 3.0, 5.0, 2.0),
        (6.0, 2.0, 5.0, 3.0),
        (5.0, 4.0, 4.0, 4.0),
    ]

    benchmark_results = []

    for i, (bL1, bL2, bx, by) in enumerate(benchmark_data, start=1):

        try:

            btheta1, btheta2 = inverse_kinematics(
                bL1,
                bL2,
                bx,
                by,
                "elbow_up"
            )

            _, _, bx_calc, by_calc = forward_kinematics(
                bL1,
                bL2,
                btheta1,
                btheta2
            )

            berr = np.sqrt(
                (bx - bx_calc)**2 +
                (by - by_calc)**2
            )

            benchmark_results.append([
                i,
                bL1,
                bL2,
                bx,
                by,
                btheta1,
                btheta2,
                bx_calc,
                by_calc,
                berr,
                "PASS" if berr < 1e-6 else "CHECK"
            ])

        except ValueError as e:

            benchmark_results.append([
                i,
                bL1,
                bL2,
                bx,
                by,
                np.nan,
                np.nan,
                np.nan,
                np.nan,
                np.nan,
                "UNREACHABLE"
            ])

    benchmark_table = pd.DataFrame(
        benchmark_results,
        columns=[
            "Test",
            "L1",
            "L2",
            "Target X",
            "Target Y",
            "Theta 1",
            "Theta 2",
            "Calculated X",
            "Calculated Y",
            "Error",
            "Status"
        ]
    )

    display(benchmark_table.round(5))


# ============================================================
# 8. ARM MOVEMENT ANIMATION
# ============================================================

print("\nCreating robot arm movement animation...")

# Movement starts from a simple configuration
start_theta1 = 0
start_theta2 = 0

frames = 80

theta1_values = np.linspace(
    start_theta1,
    theta1,
    frames
)

theta2_values = np.linspace(
    start_theta2,
    theta2,
    frames
)

fig, ax = plt.subplots(figsize=(8, 8))

limit = L1 + L2 + 1

ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)

ax.set_aspect("equal")
ax.grid(True, alpha=0.3)

ax.set_xlabel("X Position")
ax.set_ylabel("Y Position")
ax.set_title("2-Link Robot Arm Movement")

# Target
ax.scatter(
    TARGET_X,
    TARGET_Y,
    s=180,
    marker="*",
    label="Target"
)

# Arm line
arm_line, = ax.plot(
    [],
    [],
    marker="o",
    linewidth=5,
    markersize=10,
    label="Robot Arm"
)

# Path followed by end effector
path_line, = ax.plot(
    [],
    [],
    linestyle="--",
    linewidth=2,
    label="End-Effector Path"
)

ax.legend()

path_x = []
path_y = []


def update(frame):

    t1 = theta1_values[frame]
    t2 = theta2_values[frame]

    jx, jy, ex, ey = forward_kinematics(
        L1,
        L2,
        t1,
        t2
    )

    # Update robot arm
    arm_line.set_data(
        [0, jx, ex],
        [0, jy, ey]
    )

    # Update end-effector path
    path_x.append(ex)
    path_y.append(ey)

    path_line.set_data(
        path_x,
        path_y
    )

    return arm_line, path_line


animation = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=40,
    blit=True
)

plt.close(fig)

display(HTML(animation.to_jshtml()))

print("\n✓ Project completed successfully.")
print("✓ Matrix multiplication completed.")
print("✓ Rotation matrices calculated.")
print("✓ Inverse kinematics calculated.")
print("✓ Forward-kinematics verification completed.")
print("✓ Benchmark tests completed.")
print("✓ Robot-arm animation generated.")
```
